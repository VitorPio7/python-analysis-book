```python
## Expressing conditional logic as Array
```

```python
xarr = np.array([1.1, 1.2, 1.3,1.4,1.5])
```

```python
yarr = np.array([2.1, 2.2, 2.3, 2.4, 2.5])
```

```python
cond = np.array([True, False, True, True, False])
```

```python
result = np.where(cond, xarr, yarr) ## funciona como se fosse um if ternário
##O primeiro parâmetro é a condicao, e os dois ultimos são os valores, o true seria o xarr, e o false o yarr
```

```python
result
```
