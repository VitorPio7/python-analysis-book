```python
draws = rng.integers(0, 2, size=(nwalks,nsteps))##rng é um gerador de numeros aleatórios, o 0 significa o valor mínimo e o 2 o máximo, o nwalks e o nsteps representa o tamânho da matriz
```

```python
steps = np.where(draws > 0, 1, -1) ##se for maior do que zero coloca o 1, se não o -1
```

```python
walks = steps.cumsum(axis=1)
```
