```python
import re ##Remove simbolos
```

```python
states = ["Alabama", "Georgia!","Georgia","georgia","FlOrIda","south  carolina##", "West virginia?"]
```

```python
def clean_strings(strings):
    result = []
    for value in strings:
        value = value.strip()##Remove os espaços em branco iniciais
        value = re.sub("[!#?]","",value) ##(padrao, substituicao, valor)
        value = value.title()
        result.append(value)
    return result
```

```python
clean_strings(states)
```

```python
##Generator é uma maneira de lancar um output um de cada vez
```

```python
def squares(n=10):
    print(f"Generating squares from 1 to {n ** 2}")
    for i in range(1, n+1):
        yield i ** 2
```

```python
gen = squares()    
```

```python
gen
```

```python
for x in gen:
    print(x, end=" ")
```
