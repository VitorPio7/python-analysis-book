```python
import random
```

```python
import matplotlib.pyplot as plt
```

```python
import numpy as np
```

```python
position = 0
```

```python
walk = [position]
```

```python
nsteps = 1000
```

```python
for _ in range(nsteps):
    step = 1 if random.randint(0, 1) else -1
    position +=step
    walk.append(position)
```

```python
plt.plot(walk[:100])
```
