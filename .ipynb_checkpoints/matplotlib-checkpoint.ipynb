```python
import matplotlib.pyplot as plt
```

```python
%matplotlib inline
```

```python
import numpy as np
```

```python
x = np.linspace(0,5,11)
```

```python
y = x ** 2
```

```python
x
```

```python
y
```

```python
plt.plot(x,y)
plt.xlabel('X Label')
plt.ylabel('Y Label')
plt.title('Title')
```

```python
plt.subplot(1,2,1)
plt.plot(x,y,'r')
plt.subplot(1,2,2)
plt.plot(y,x,'b')
```

```python
## OO
## É mais personalizável do que o anterior, mas chegamos nos mesmos resultados
fig = plt.figure()

axes = fig.add_axes([0.1,0.1,0.8,0.8])

axes.plot(x,y)
axes.set_xlabel('X Label')
axes.set_ylabel('Y Label')
axes.set_title('Set Title')
```

```python
fig = plt.figure()

axes1 = fig.add_axes([0.1,0.1,0.8,0.8])
axes1.plot(x,y)

```

```python
fig, axes = plt.subplots(nrows=1, ncols=2) ##Não precisamos adicionar o axis, aqui ele coloca automaticamente

##for current_ax in axes:
##  current_ax.plot(x,y)

axes[0].plot(x,y)
axes[0].set_title('First Plot')

axes[1].plot(y,x)
axes[1].set_title('Second Plot')
```

```python
fig = plt.figure()

ax = fig.add_axes([0,0,1,1])
ax.plot(x, x,label='X Squared') ## label vai implementar a consulta da tabela
ax.plot(x, x**2,label='X Cubed')
ax.set_title('Title')
ax.set_ylabel('Y')
ax.set_xlabel('X')
ax.legend(loc=0)
```

```python
fig = plt.figure()

ax = fig.add_axes([0,0,1,1])

ax.plot(x, y, color="purple",linewidth=1, alpha=1, linestyle='-', marker='o', markersize=8,
       markerfacecolor="yellow",markeredgewidth=3,markeredgecolor='green') 
## linewidth diz sobre a espessura, alpha diz sobre a opacidade
ax.set_xlim([0,1]) ##Permite traçar ou isolar os dados
ax.set_ylim([0,2]) ##Permite isolar no eixo y
```

```python
fig, axes = plt.subplots(1, 2, figsize=(10,4)) ## 1 linha e duas colunas do tamanho 10x4
axes[0].plot(x,x**2,x,np.exp(x)) ## Incersão dos dados na tabela
axes[0].set_title("Normal scale") ## coloca o título do gráfico
axes[1].plot(x, x**2,x,np.exp(x)) ## coloca info no segundo gráfico
axes[1].set_yscale("log") ## muda o tipo da escala de y
axes[1].set_title('Logarithmic scale(y)')
```

```python
fig, ax = plt.subplots(figsize=(10,4))
ax.plot(x,x**2,x,x**3,lw=2)

ax.set_xticks([1, 2, 3, 4, 5])
ax.set_xticklabels([r'$\alpha$', r'$\beta$', r'$\gamma$', r'$\delta$', r'$\epsilon$'], fontsize=18)
```
