```python
x = df[['Avg. Area Income', 'Avg. Area House Age', 'Avg. Area Number of Rooms',
       'Avg. Area Number of Bedrooms', 'Area Population', 'Price', 'Address']]
```

```python
X_train, X_test, y_train, y_test = train_test_split( x, y, test_size=0.4, random_state=101) ## nessa linha de codigo vamos testar e treinar os dados
```

```python
from sklearn.linear_model import LinearRegression
```

```python
X_train = X_train.select_dtypes(include=['number']) ##utilizacao do linear regression
regressor = LinearRegression().fit(X_train,y_train) 
regressor.get_params()
```

```python
print(regressor.intercept_)
```

```python
regressor.coef_
```
