```python
fig = plt.figure()

ax = fig.add_axes([0,0,1,1])
ax.plot(x, x,label='X Squared') ## label vai implementar a consulta da tabela
ax.plot(x, x**2,label='X Cubed')
ax.set_title('Title')
ax.set_ylabel('Y')
ax.set_xlabel('X')
ax.legend(loc=0)
```

```python
fig = plt.figure()

ax = fig.add_axes([0,0,1,1])

ax.plot(x, y, color="purple",linewidth=1, alpha=1, linestyle='-', marker='o', markersize=8,
       markerfacecolor="yellow",markeredgewidth=3,markeredgecolor='green') 
## linewidth diz sobre a espessura, alpha diz sobre a opacidade
ax.set_xlim([0,1]) ##Permite traçar ou isolar os dados
ax.set_ylim([0,2]) ##Permite isolar no eixo y
```

```python
fig, axes = plt.subplots(1, 2, figsize=(10,4)) ## 1 linha e duas colunas do tamanho 10x4
axes[0].plot(x,x**2,x,np.exp(x)) ## Incersão dos dados na tabela
axes[0].set_title("Normal scale") ## coloca o título do gráfico
axes[1].plot(x, x**2,x,np.exp(x)) ## coloca info no segundo gráfico
axes[1].set_yscale("log") ## muda o tipo da escala de y
axes[1].set_title('Logarithmic scale(y)')
```

```python
fig, ax = plt.subplots(figsize=(10,4)) ## Criação de gráfico
ax.plot(x,x**2,x,x**3,lw=2)

ax.set_xticks([1, 2, 3, 4, 5]) ## Customização do nosso eixo x
ax.set_xticklabels([r'$\alpha$', r'$\beta$', r'$\gamma$', r'$\delta$', r'$\epsilon$'], fontsize=18)

yticks = [0, 50, 100, 150] ## Eixo y customizado
ax.set_yticks(yticks)
```

```python
fig, ax = plt.subplots(1, 1)

ax.plot(x, x**2, x, np.exp(x))
ax.set_title("scientific notation")

ax.set_yticks([0, 50, 100, 150])

from matplotlib import ticker
formatter = ticker.ScalarFormatter(useMathText=True)
formatter.set_scientific(True)
formatter.set_powerlimits((-1,1))
ax.yaxis.set_major_formatter(formatter)
```

```python
plt.rcParams['xtick.major.pad'] = 5
plt.rcParams['ytick.major.pad'] = 5

fig, ax = plt.subplots(1,1)

ax.plot(x, x**2,x,np.exp(x))
ax.set_yticks([0, 50, 100, 150])

ax.set_title("label and axis spacing")
ax.yaxis.labelpad = 5
ax.yaxis.labelpad = 5

ax.set_xlabel("x")
ax.set_ylabel("y")

fig.subplots_adjust(left=0.15, right=.95, bottom=.1,top=.95) ## Ajusta o tamanho do gráfico
ax.grid(color='gray',alpha=0.5,linestyle='dashed',linewidth=0.5) ##Ajusta e configura o grid
ax.spines['bottom'].set_color('red') ## Altera a cor da linha abaixo

ax2 = ax.twinx()
ax2.plot(x, x**3, lw=2, color="blue") ##Criamos uma reta irmã com essas configuracões
ax2.set_ylabel(r"volume $(m³)$")
for label in ax2.get_yticklabels():
    label.set_color("red")
```

```python
fig, ax = plt.subplots()

ax.spines['right'].set_color('black')
ax.spines['left'].set_color('red')
ax.xaxis.set_ticks_position('bottom')
ax.spines['bottom'].set_position(('data',0))
ax.yaxis.set_ticks_position('left')
ax.spines['left'].set_position(('data',0))

xx = np.linspace(-0.75,1.,100)
ax.plot(xx,xx**3)
```
